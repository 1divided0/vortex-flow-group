# 2d Gitter [Python]

In Python lässt sich ein zweidimensionales Gitter mit `numpy.meshgrid` erzeugen.

Weitere Informationen dazu finden sich in der Dokumentation <https://numpy.org/doc/stable/reference/generated/numpy.meshgrid.html>.

![2d Gitter Python](../.Dateien/Bilder/2d_Gitter_Python.svg)

```python
import numpy as np

m = 5
n = 3

x = np.linspace(0,1,m)
y = np.linspace(0,1,n)

X,Y = np.meshgrid(x,y)
```

Hier wurde die folgende Produktmenge erstellt:

$$ \underbrace{\{0.0,0.5,1.0\}}_\text{y-Werte}\times\underbrace{\{0.0,0.25,0.5,0.75,1.0\}}_\text{x-Werte} $$

Die Werte wurden genau in dem Raster in zwei separaten Matrizen abgespeichert.
