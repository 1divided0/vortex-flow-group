# Vektorisierung [Python]

In Python lässt sich eine Matrix mit [`numpy.reshape`](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) vektorisieren. Normalerweise geschieht dies durch Verkettung der Zeilen, weil in einem Array die Daten geschachtelt werden und deren Struktur so erhalten bleibt.

$$\begin{align*} \forall i\in\{0,\ldots,m-1\},\forall j\in\{0,\ldots,n-1\}&\colon\\\\ (j,i) \mapsto&~ jm+i \end{align*}$$

![Verkettung Python](../.Dateien/Bilder/Verkettung_Python.svg)

```python
import numpy as np

m = 3
n = 2

A = np.ones((n,m))
a = np.reshape(A,(n*m))
```

Hier wurde eine `2×3`-Matrix in einen Vektor überführt.

```python
A
```

```python
a
```

Diese Operation lässt sich auch umkehren.

```python
B = np.reshape(a,(n,m))
B
```
