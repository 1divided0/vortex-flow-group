# Vektorisierung [Matlab]

In Matlab lässt sich eine Matrix mit [`reshape`](https://www.mathworks.com/help/matlab/ref/reshape.html) vektorisieren. Normalerweise geschieht dies durch Verkettung der Spalten, weil bei einer Matrix-Vektor Multiplikation in den Spaltenraum der Matrix abgebildet wird und es auf diese Weise natürlich erscheint, dessen Topologie beizubehalten.

\begin{align*} \forall i\in\{1,\ldots,m\},\forall j\in\{1,\ldots,n\}&\colon\\\\ (j,i) \mapsto&~ (i-1)n+j \end{align*}

![Verkettung Python](../.Dateien/Bilder/Verkettung_Matlab.svg)

```matlab
m = 3;
n = 2;

A = ones(n,m);
a = A(:);
```

Hier wurde eine `2×3`-Matrix in einen Spaltenvektor überführt.

```matlab
A
```

```matlab
a
```

Diese Operation lässt sich auch umkehren.

```matlab
B = reshape(a,[n m])
```
