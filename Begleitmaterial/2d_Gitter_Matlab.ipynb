# 2d Gitter [Matlab]

In Matlab lässt sich ein zweidimensionales Gitter mit [`meshgrid`](https://www.mathworks.com/help/matlab/ref/meshgrid.html) erzeugen.

![2d Gitter Matlab](../.Dateien/Bilder/2d_Gitter_Matlab.svg)

```matlab
m = 5;
n = 3;

x = linspace(0,1,m);
y = linspace(0,1,n);

[X,Y] = meshgrid(x,y);
```

Hier wurde die folgende Produktmenge erstellt:

$$ \underbrace{\{0.0,0.5,1.0\}}_\text{y-Werte}\times\underbrace{\{0.0,0.25,0.5,0.75,1.0\}}_\text{x-Werte} $$

Die Werte wurden genau in dem Raster in zwei separaten Matrizen abgespeichert.
